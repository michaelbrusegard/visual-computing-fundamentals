```python
from utils import read_im, save_im
import matplotlib.pyplot as plt
import numpy as np

im = read_im("results/duck_greyscale.jpeg")

def inverse(im):
    height = len(im)
    width = len(im[0])

    inverted = [[0 for _ in range(width)] for _ in range(height)]

    for i in range(height):
        for j in range(width):
            inverted[i][j] = 255 * (1 - im[i][j] / 255)

    return np.array(inverted)

im_inverse = inverse(im)
im_inverse = np.clip(im_inverse, 0, 255)
im_inverse = im_inverse.astype(np.uint8) 

plt.imshow(im_inverse, cmap='gray')
save_im("results/duck_inverse.jpeg", im_inverse, cmap='gray')
```
